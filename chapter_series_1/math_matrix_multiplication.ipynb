# Matrix Multiplication

```python
# imports
import numpy as np
import torch
```

## Numpy

```python
# create matrix
mat_1 = np.random.randn(2,3)
mat_2 = np.random.randn(3,4)
print(mat_1,end='\n\n\n')
print(mat_2,end='\n\n\n')
mat_mul = np.round(mat_1@mat_2, 2) #alternative way: np.matmul(mat_1, mat_2)
print(mat_mul)
```

## pyTorch

```python
mat_a = torch.randn(3,3)
mat_b = torch.randn(3,6)
mat_c = torch.randn(5,3)
mat_c1 = torch.randn(7,3)

print(np.round(mat_a@mat_b,2))
print(np.round(mat_c@mat_c1.T,2))
print(np.round(mat_c1@mat_a,2))
```
