```python
# imports
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
```

```python
# Params
n1 = 30 # samples in dataset 1
n2 = 30 #sample in dataset 2


#generate datasets
data1 =  [1,-4,3,6,7,4,6,3,2,1,7,8,9,6,5,0,-1,4,0,1,1,0,3,4,6,3,-1,3,2,9,3]
data2 =  [0,2,-1,9,0,1,8,9,-2,5,4,5,2,8,4,5,6,5,2,1,9,0,0,0,0,1,2,3,2,8,1]
mean1 = np.mean(data1)# population mean in dataset1
mean2 = np.mean(data2) # population mean in dataset2
print(mean1,mean2)
```

```python
# plotting data
plt.rcParams.update({'font.size': 12})
plt.plot(0 + np.random.randn(n1)/13,data1,'wo',markerfacecolor='#BFC6C4',markersize=10)
plt.plot(1  + np.random.randn(n2)/13,data2,'go',markerfacecolor='#73AF6F',markersize=10)
plt.xlim([-1,2])
plt.xticks([0,1],labels=['Group A','Group B'])
plt.show()
```

```python
# t-test via stat package
# _ind = independent samples
t,p = st.ttest_ind(data1,data2)
print(t)
print(p)
```
