```python
from time import time

import numpy as np
import torch
```

```python
# ! Time Ms
time_ms_64 =lambda : np.int64(time()*10)
time_ms_16 =lambda : np.int16(time_ms_64()%1000)
```

# Numpy Random Seed

```python
# generate array of random numbers
arr = np.random.randn(3)
print(arr)
```

```python
#  ? Old Seeding mechanism
np.random.seed(3)
print(np.random.random(3))
print(np.random.random(3))
# [0.5507979  0.70814782 0.29090474]
# [0.51082761 0.89294695 0.89629309]
```

```python
# ? another technique using old method
np.random.seed(time_ms_16())
print(np.random.random(3))
print(np.random.random(3))
```

```python
# * New Seeding Mechanism
rand_seed = np.random.RandomState(time_ms_16())

print(rand_seed.randn(3))
```

# PyTorch Random Seed

```python
torch.randn(5)
```

```python
# Fix seeding
torch.manual_seed(15)
print(torch.randn(5))
```

```python
# dynamic seeding
torch.manual_seed(time_ms_16())
print(torch.randn(5))
```
