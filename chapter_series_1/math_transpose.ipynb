```python
#import libraries
import numpy as np
import torch
```

# NUMPY

```python
# creating a vector
vec = np.array([ [1,2,3,4] ])
print(vec)

vec_2d = np.array([ ['a','b','c','d'], ['e','f','g','h'] ])
print(vec_2d)
```

## Transposing a vector using Numpy

```python
t_vec = np.transpose(vec) # RETURNS TRANSPOSED COPY OF THE VECTOR USING np.transpose
print(t_vec)

print(vec.T) # RETURNS TRANSPOSED COPY OF THE VECTOR USING vector.T
```

## WORKING WITH 2D Matrix

```python
transposed_vec_2d = vec_2d.T
print(transposed_vec_2d)
```

# pyTORCH

```python
t_vec = torch.tensor([[1,2,3,4]])
t_vec_2d = torch.tensor([[1,2,3,4], [5,6,7,8] ])
print(t_vec)
print(t_vec_2d)
```

## Transposing vector using pyTorch

```python
print(t_vec.T)
print(t_vec_2d.T)
```
