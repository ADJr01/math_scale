```python
# imports
import numpy as np
import torch
```

# Mean Calculation

```python
# Dataset
sample_data = [168,170,150,160,182,145,175]
n=len(sample_data)
```

```python
mean1 = np.mean(sample_data)
mean2 = np.sum(sample_data)/n
print(mean1)
print(mean2)
```

# Variance Calculation

```python
v = np.var(sample_data,ddof=1)
v1 = (1/(n-1)) * np.sum((sample_data-mean1)**2)
print(v)
print(v1)
```
