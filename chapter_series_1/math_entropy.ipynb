<h1 align="start">Entropy Formula</h1>
<h2>$$H(X) = -\sum_{i=1}^{n} P(x_i) \log_b P(x_i)$$</h2>

```python
# probability of event happening
p=0.40

# Not Correct Formula
H = -(p*np.log(p)) # we need to calculate it on a dataset
print(f"Wrong Entropy: {H}")
```

```python
# short function to calculate entropy
def calc_entropy(probabilities)->float:
    H:float = 0
    for p in probabilities:
        H += -(p*np.log(p))
    return H

def calc_cross_entropy(probabilities,truth_series)->float:
    if len(probabilities)!=len(truth_series):
        raise Exception("Length of probabilities and truth series do not match")
    H:float=0.0
    for index,p in enumerate(probabilities):
        q = truth_series[index]
        H+=-(p*np.log(q))
    return H
```

```python
x = [0.25,0.75]
H = calc_entropy(probabilities=x)
print(f"Correct Entropy: {H}")
```

### Binary-Cross-Entropy

```python
p_1 = x[0]
p_2 = x[1]

binary_cross_entropy = -(p_1*np.log(p_1) + (1-p_1)*np.log(1-p_1))
print(f"Binary Cross Entropy: {binary_cross_entropy}")
```

## Cross Entropy

```python
p = [1,0.01] # Truth
q = [0.25,0.75] # Predicted
cross_entropy = calc_cross_entropy(probabilities=p,truth_series=q)
print(f"Cross Entropy: {cross_entropy}")
```
