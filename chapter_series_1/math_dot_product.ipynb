# Dot Product

```python
import numpy as np
import torch
```

```python
#create two vector
vec_1 = np.array([0.5,0.25,0,-0.16])
vec_2 = np.array([0.1,0,0.31,-0.8])
dot_prod = np.dot(vec_1, vec_2) # dot product via np.dot function
print(dot_prod)

print(np.sum(vec_1*vec_2)) # dot product via np.sum(vector1*vector2)

# creating two matrix
mat_1 = np.array([[1,2,3,4,5],[6,-1,0,3,2]])
mat_2 = np.array([[0,0,-1,-4,3],[0,9,1,0,5]])
mat1_dot = np.dot(mat_1, mat_1.T)
mat2_dot = np.dot(mat_2, mat_2.T)
print(f"mat1_dot: {mat1_dot}")
print(f"mat2_dot: {mat2_dot}")
```

# pyTorch

```python
t_vec1 = torch.tensor([0.5,0.23,0.3])
t_vec2 = torch.tensor([0.1,-0.15,0.87])
t_dot_product = torch.dot(t_vec1,t_vec2)
print(f"t_dot_product: {t_dot_product}")
```
