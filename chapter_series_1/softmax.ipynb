```python
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

```python
# input list
z_list = [1,2,3]
```

### Manual Implementation

#### NumPy

```python
num = np.exp(z_list)
den = np.sum(num)
sigma = num/den
print(sigma)
```

#### pyTorch

```python
soft_max = nn.Softmax(dim=0)
sigmaT = soft_max(torch.Tensor(z_list))
print(sigmaT)
```
